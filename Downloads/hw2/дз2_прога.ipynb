## ЧАСТЬ 3: создание и преобразование столбцов (1 балл)

*Создайте колонку 'Top_University' со значением True для университетов, входящих в топ-50*

```python
df['ranking_global'] = pd.to_numeric(df['ranking_global'], errors='coerce')
df['Top_University'] = df['ranking_global'] <= 50
df
```

*Создайте колонку Academic_Performance, в которую внесите значение:*
- *"High", если GPA 3.5 баллов или выше,*
- *"Medium", если GPA от 3 до 3.49 баллов,*
- *"Low", если GPA ниже 3 баллов,*
- *Если GPA неизвестен, в этой колонке также должен быть пропуск.*

```python
def get_performance(gpa):
    if pd.isna(gpa):
        return pd.NA
    elif gpa >= 3.5:
        return 'High'
    elif 3.0 <= gpa < 3.5:
        return 'Medium'
    else:
        return 'Low'

df['Academic_Performance'] = df['GPA'].apply(get_performance)
df
```

*Определите топ-5 стран с самыми высокими значениями GPA (средним по стране).*

```python
df['GPA'] = pd.to_numeric(df['GPA'], errors='coerce')
top1 = df.groupby('Country')['GPA'].mean().sort_values(ascending=False).head(5)
print(top1)
```

*Сгруппируйте записи по профессиям и посчитайте:*
- *количество людей в каждой профессии,*
- *средний GPA для каждой профессии.*

```python
df['GPA'] = pd.to_numeric(df['GPA'], errors='coerce')
profession = df.groupby('Occupation')['GPA'].agg(
    People_Count='count',
    Average_GPA='mean'
).reset_index()
profession = profession.sort_values(by='People_Count', ascending=False)
print(profession)
```
